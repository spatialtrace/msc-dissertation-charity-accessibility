# Distance-based E2SFCA accessibility, 2001/2011/2021

This is an isolated, executable dissertation workflow for the fixed internal set of 3,411 2021 LSOAs.

**Workflow:** Data collection → common spatial foundation → OD/network impedance → E2SFCA → A/A* standardisation → descriptive outputs → longitudinal analysis preparation.

- Demand: Care50 count.
- Capacity: charity-level `log1p(income_2021_gbp)`, then provider-LSOA aggregation.
- Impedance: 20 km road distance.
- Bands: 0–5 / >5–10 / >10–20 km, weights `1.00 / 0.68 / 0.22`.
- External LSOAs are boundary-correction support only and never enter final result rows.
- Buffer membership creates candidates only. Every E2SFCA pair requires a finite graph path, both connectors and the true catchment threshold.
- Trajectory analysis is deliberately not run in this notebook.

## Reproducibility contract

**Purpose.** Repeat the accessibility calculation with a 20 km road-distance impedance specification while keeping the common geography, demand, capacity, boundary support and A* benchmark rules fixed.

**Interpretation.** This is a specification sensitivity analysis, not an alternative main model. Licensed road data, graphs and OD caches are intentionally excluded from the public repository.
